```python
import requests

url = "https://maze.edwarddonner.com/"
response = requests.get(url)

print(response.text)
```

```python
def open_webpage(url):
    """Open a webpage and return its text content."""
    response = requests.get(url)
    response.raise_for_status()
    return response.text
```

```python
room = open_webpage("https://maze.edwarddonner.com/")

print(room[:3000])
```
